### defaultdict

`defaultdict` из библиотеки ` collections` - это словарь, у которого на все ключи задано значение по умолчанию.

```python
from collections import defaultdict

wordcount = defaultdict(int)  # тип данных, по которому берется дефолтное значение
words = input().split()
for word in words:
    wordcount[word] += 1

print(wordcount)
```

### Итерируемый объект

**Итерация** - это взятие элементов по очереди.


**Итератор** — это объект, который возвращает свои элементы по одному за раз. Одно из главных преимуществ итераторов — то, что он не хранит все свои элементы в памяти, а вычисляет их "на ходу".

У всех итераторов есть метод __next()__, который возвращает следующий элемент, если он есть, и вызывает исключение StopIteration, если элементов не осталось. Метод можно вызвать, передав итератор в функцию next.

**Итерируемый объект** — объект, который можно итерировать, т.е. возвращать элементы по одному. Мы уже встречались итерируемыми объектами: списки, кортежи, строки, множества — итерируемые. По их элементам можно пройтись с помощью цикла for.

Важно понимать разницу между понятиями *итератор* и *итерируемый объект*. Списки, кортежи, строки, множества являются итерируемыми объектами (по их элементам можно пройтись с помощью списка for), но не являются итераторами и не реализуют метод __next()__. При этом на их основе можно создать итератор с помощью функции iter, который реализует метод __iter__().

Примеры итерируемых объектов:

* список
* строка
* кортеж
* словари
* множества
* файлы (они будут в следующих занятиях)

Когда мы создаем цикл for, наш объект проходит именно по итератору.
Действие ниже:

for object in iterable:

    действие
    
вызывает метод итерируемого объекта , который возвращает итератор.
Таким образом, создаем объект-итератор , по которому и бежит цикл for.

Для того чтобы все это увидеть, есть функция iter().

 В качестве аргумента ей передается итерируемый объект (словарь, список, кортеж и т.д.) , а она возвращает соответствующий итератор.


```python
collection = [1, 2, 3, 4, 5]
# def __iter__ - метод __iter()__ возвращает итератор, метод __next()__ — следующий элемент потока данных.
list_iter = iter(collection)

for item in list_iter:
  print(item)
```

Заметим, что если мы попытаемся пройти по итератору еще раз, элементов в нём уже не будет:

Если элементов не осталось, возникает исключение `StopIteration     `

В коде выше во второй строке создается итератор на основе списка lst. Функция next каждый раз возвращает очередной элемент, пока они не закончатся; при этом если элементы закончились, очередной вызов next выбросит исключение StopIteration.

У итераторов также есть метод __iter__(), который возвращает сам итератор. Поэтому итераторы также являются итерируемыми объектами.

Рассмотрим код ниже:

```python
lst = [4, 12, 11, 5, 7]
lst_iterator = iter(lst)

for number in lst_iterator:
    print(number)

for number in lst_iterator:
    print(number)
```

```python
next(lst_iterator )
```

В коде происходит следующее:

1. Создаем список lst.

2. lst_iterator = iter(lst) — создается итератор на основе списка lst

3. Первый цикл for: внутри цикла каждый элемент итератора по очереди присваивается переменной number, и print(number) выводит этот элемент.

4. Второй цикл for также пытается пройтись по lst_iterator, но важно понимать: после того, как итератор достиг конца в первом цикле, у него больше нет элементов. Поэтому второй цикл не выводит ничего.



У самого списка метода next нет! Когда мы проходимся по элементам списка с помощью цикла for, "под капотом" на основе списка создается итератор, который и возвращает элементы по одному, пока они не закончатся. Каждую итерацию цикла у итератора вызывается метод __next__(), а при получении исключения StopIteration цикл останавливается.

Например, код ниже:

```python
lst = [4, 12, 11, 5, 7]
for number in lst:
    print(number)
for number in lst:
    print(number)
```

будет работать примерно так:

```python
lst = [4, 12, 11, 5, 7]
lst_iterator = iter(lst)
while True:
    try: # код, который будет выполняться, если всё хорошо
        print(next(lst_iterator))
    except StopIteration: # код, который будет выполняться, если произошла ошибка StopIteration (так мы "ловим" это ошибку; подробнее про это поговорим в следующих занятиях)
        break
```

**4. Множества (set)**

Множество — это неупорядоченная коллекция уникальных элементов. Но несмотря на то, что порядок в множествах не определен, они все равно итерируемые, и их элементы можно перебрать (правда, порядок может быть произвольным).

```python
set_numbers = {1, 3, 7}
for elem in set_numbers:
    print(elem)
```

Но у множеств есть особенность, связанная с их неупорядоченностью и внутренней структурой: если попытаться изменить множество во время итерации, попытка выдать следующий элемент сразу приведет к ошибке:

```python
set_numbers = {1, 2, 7}
for elem in set_numbers:
    set_numbers.add(8)
    print(elem)
```

```python
set_numbers = {1, 2, 7}
iter_set = iter(set_numbers)
```

```python
for elem in iter_set:
  set_numbers.add(8)
  print(elem)
```

## Генераторы списков (List Comprehensions)
Это способ быстро создать новый список, не применяя цикл for.

```python
lst = [i**2 for i in range(20)]

print(type(lst))
print(lst)
```

```python
lst = list()
for i in range(20):
    lst.append(i**2)
```

```python
lst
```

Это более короткая запись кода ниже:

```python
lst = list()
for i in range(20):
    lst.append(i**2)
print(lst)
```

Общая конструкция выглядит так:

**new_list** = [**expression** for **member** in **iterable**]

1. **expression** - какое-либо выражение (например, переменная или вызов функции), которое возвращает значение.
2. **member** является объектом или значением в списке или итерируемым объекте (iterable).
3. **iterable** список, множество, последовательность, генератор или любой другой объект, который может возвращать свои элементы по одному.


 В list comprehension можно добавить какое-нибудь условие (как мы это делали с **if**). Выглядеть это будет так:

**new_list** = [**expression** for **member** in **iterable** (if **conditional**)]

Разберем на примере:

```python
lst = [5, 1, 2, 54, 3, 7, 89]
filter_lst = [i for i in lst if i % 2 == 0]
print(filter_lst)
```

```python
filter_lst = list()
for i in lst:
    if i % 2 == 0:
        filter_lst.append(i)
```

```python
filter_lst
```

Также можно использовать разное значение выражения в зависимости от условия:
**new_list** = [**expression** (if **conditional**) for **member** in **iterable**]

```python
original_prices = [1.25, -9.45, 10.22, 3.78, -5.92, 1.16]
prices = [i if i > 0 else 0 for i in original_prices]
prices
```

```python
prices = list()
for i in original_prices:
    if i > 0:
        prices.append(i)
    else:
        prices.append(0)
```

```python
prices
```

Хотя именно **list comprehension** в Python является распространенным инструментом, вы также можете создавать множественные и словарные представления (**set and dictionary comprehensions**).


**set comprehension** задается почти точно так же, как  **list comprehension**.

Разница в том, что выходные данные не содержат дубликатов.

```python
quote = "This is the text in which you need to find vowels"
unique_vowels = {letter for letter in quote if letter in 'aeiou'}
unique_vowels
```

```python
quote = "This is the text in which you need to find vowels"
our_set = set()
for letter in quote:
  if letter in  'aeiou':
    our_set.add(letter)
```

```python
 our_set
```

**Dictionary comprehensions** , по сути, работает так же, но с дополнительным требованием определения ключа. Ключ отделяется двоеточием.


```python
squares = {i: i**2 for i in range(10)}
squares
```

```python
squares = {}
for i in range(10):
  squares[i] = i * i
print(squares)
```
